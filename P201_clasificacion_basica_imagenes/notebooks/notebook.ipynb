```python
import json
import pickle
from pathlib import Path

from sklearn import datasets
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split

cwd = Path.cwd()
if (cwd / "submission").is_dir():
    SUBMISSION_DIR = cwd / "submission"
elif (cwd.parent / "submission").is_dir():
    SUBMISSION_DIR = cwd.parent / "submission"
else:
    SUBMISSION_DIR = cwd / "P201_clasificacion_basica_imagenes" / "submission"

digits = datasets.load_digits()
n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))

X_train, X_test, y_train, y_test = train_test_split(
    data,
    digits.target,
    test_size=0.5,
    random_state=0,
    stratify=digits.target,
)

estimator = LogisticRegression(max_iter=1000)
estimator.fit(X_train, y_train)

predictions = estimator.predict(X_test)
predicted_proba = estimator.predict_proba(X_test)

print("Accuracy train:", accuracy_score(y_true=y_train, y_pred=estimator.predict(X_train)))
print("Accuracy test:", accuracy_score(y_true=y_test, y_pred=predictions))
print("Confusion matrix:\n", confusion_matrix(y_test, predictions))

SUBMISSION_DIR.mkdir(parents=True, exist_ok=True)

with open(SUBMISSION_DIR / "estimator.pkl", "wb") as file:
    pickle.dump(estimator, file)

metrics = {
    "test_accuracy": float(accuracy_score(y_true=y_test, y_pred=predictions)),
    "test_size": int(len(y_test)),
}
with open(SUBMISSION_DIR / "metrics.json", "w", encoding="utf-8") as file:
    json.dump(metrics, file, indent=2)

print("Archivos guardados en:", SUBMISSION_DIR)
```
